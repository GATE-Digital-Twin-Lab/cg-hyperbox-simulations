# Intersected Intermediate CG-Projection Enclosure

This is the v4 counterpart of `system_303_solver_v3_simplified.ipynb`.
It retains v3's data, boundary-coordinate removal, Jacobi scaling, 520-step
CG trajectory, and stable prefix-direction selector.

For every $j=1,\ldots,m$, the notebook constructs the raw projected hull
$B_j$ and updates

$$C_j=C_{j-1}\cap B_j,\qquad C_0=B_0.$$

Only the stored animation snapshots are decimated: $C_{10},C_{20},\ldots$
are retained. This is deliberately different from computing only
$B_0\cap B_{10}$, $B_0\cap B_{20}$, and so on.


```python
from pathlib import Path
from time import perf_counter

import matplotlib.pyplot as plt
import numpy as np
from IPython.display import HTML, display
from matplotlib.animation import FuncAnimation
from scipy.linalg import qr, solve
from scipy.sparse.linalg._isolve.utils import make_system


data_dir = Path("system_303")
A_full = np.genfromtxt(data_dir / "stiffness_beam_A.dat", delimiter=",")
b_full = np.genfromtxt(data_dir / "forcing_b.dat")
box = np.genfromtxt(data_dir / "B0_wide_bc.dat", delimiter=",")
x_star_full = np.genfromtxt(data_dir / "soln_x.dat")

# Remove the first three boundary-condition coordinates, exactly as in v3.
A = A_full[3:, 3:]
b = b_full[3:]
ell0 = box[3:, 0]
u0 = box[3:, 1]
x_star = x_star_full[3:]
print(f"system dimension: {len(b)}")
```

## Jacobi scaling

With `scale = sqrt(diag(A))`, solve in coordinates $y=\mathrm{scale}\,x$
and transform all reported enclosures back to the original coordinates.
The fixed iteration horizon remains 520, exactly as in the simplified v3
experiment, so the final raw-prefix result is directly comparable with v3.


```python
diagonal = np.diag(A)
if np.any(diagonal <= 0):
    raise ValueError("Jacobi scaling requires a positive diagonal.")

scale = np.sqrt(diagonal)
A_hat = A / scale[:, None] / scale[None, :]
b_hat = b / scale
ell_hat = scale * ell0
u_hat = scale * u0

cg_projection_iterations = 520
snapshot_period = 10
projection_qr_tolerance = 1e-10
projection_max_gram_condition = 1e12
```

## SciPy CG recurrence with preserved directions

This is SciPy's unpreconditioned CG loop on the already scaled system. The
recurrence is unchanged; only the search directions, their matrix images,
and residual norms are retained. The tolerance exit is omitted so exactly
$m$ directions are produced, as required by the algorithm.


```python
def scipy_cg_with_directions(A, b, x0, maxiter):
    """SciPy's unpreconditioned CG loop, with p_k and A p_k retained.

    The recurrence is copied from scipy.sparse.linalg.cg. Its tolerance exit
    is intentionally omitted because the enclosure experiment prescribes a
    fixed number of iterations. Jacobi scaling has already been applied.
    """
    A, M, x, b, postprocess = make_system(A, None, x0, b)
    n = len(b)
    if maxiter < 1:
        raise ValueError("maxiter must be positive.")

    dotprod = np.vdot if np.iscomplexobj(x) else np.dot
    matvec = A.matvec
    psolve = M.matvec
    r = b - matvec(x) if x.any() else b.copy()

    directions = np.empty((n, maxiter), dtype=x.dtype)
    applied_directions = np.empty((n, maxiter), dtype=x.dtype)
    residual_norms = np.empty(maxiter, dtype=float)

    # This is SciPy's CG recurrence. The extra assignments only retain p,
    # A p, and the residual norm for the enclosure phase.
    rho_prev, p = None, None
    for iteration in range(maxiter):
        z = psolve(r)
        rho_cur = dotprod(r, z)
        if iteration > 0:
            beta = rho_cur / rho_prev
            p *= beta
            p += z
        else:
            p = np.empty_like(r)
            p[:] = z[:]

        directions[:, iteration] = p
        q = matvec(p)
        applied_directions[:, iteration] = q
        alpha = rho_cur / dotprod(p, q)
        x += alpha * p
        r -= alpha * q
        rho_prev = rho_cur
        residual_norms[iteration] = np.linalg.norm(r)

    return {
        "x_cg": postprocess(x),
        "residual": r,
        "residual_norms": residual_norms,
        "directions": directions,
        "applied_directions": applied_directions,
    }
```

```python
def select_independent_directions(
    directions,
    applied_directions,
    qr_tolerance=1e-10,
    max_gram_condition=1e12,
):
    """A-normalize and retain a stable pivoted-QR subset."""
    a_norm_squared = np.sum(directions * applied_directions, axis=0)
    valid = np.isfinite(a_norm_squared) & (a_norm_squared > 0)
    valid_indices = np.flatnonzero(valid)
    if valid_indices.size == 0:
        raise RuntimeError("CG produced no direction with a positive A-norm.")

    a_norm = np.sqrt(a_norm_squared[valid])
    normalized = directions[:, valid] / a_norm[None, :]
    normalized_applied = applied_directions[:, valid] / a_norm[None, :]
    _, triangular, pivots = qr(normalized, mode="economic", pivoting=True)
    diagonal = np.abs(np.diag(triangular))
    candidate_rank = int(np.sum(diagonal > qr_tolerance * diagonal[0]))

    for rank in range(candidate_rank, 0, -1):
        local = pivots[:rank]
        basis = normalized[:, local]
        applied_basis = normalized_applied[:, local]
        gram = basis.T @ applied_basis
        gram = 0.5 * (gram + gram.T)
        condition = np.linalg.cond(gram)
        if np.isfinite(condition) and condition <= max_gram_condition:
            return {
                "selected": valid_indices[local],
                "basis": basis,
                "applied_basis": applied_basis,
                "qr_rank": candidate_rank,
                "retained_rank": rank,
                "gram_condition": float(condition),
            }
    raise RuntimeError("No stable independent direction subset was found.")
```

```python
# Every prefix is selected independently below, just as in v3. This ensures
# that the raw B_j at a stored iteration agrees with v3's prefix calculation.
```

## Every-j running intersection

The expensive projection is evaluated for every integer $j$, not merely for
animation frames. Each prefix performs the same A-normalization, pivoted-QR
selection, Gram solve, and exact hull calculation as simplified v3. The
animation arrays are appended only when $j$ is a multiple of 10.


```python
def project_prefix_enclosure(A, b, lower, upper, directions, applied_directions):
    """Return the raw projected hull B_j for one complete CG prefix."""
    center = 0.5 * (lower + upper)
    initial_radius = 0.5 * (upper - lower)
    selection = select_independent_directions(
        directions,
        applied_directions,
        qr_tolerance=projection_qr_tolerance,
        max_gram_condition=projection_max_gram_condition,
    )
    basis = selection["basis"]
    applied_basis = selection["applied_basis"]
    gram = basis.T @ applied_basis
    gram = 0.5 * (gram + gram.T)

    coefficients = solve(
        gram,
        basis.T @ (b - A @ center),
        assume_a="sym",
        check_finite=False,
    )
    projected_center = center + basis @ coefficients
    gram_inverse_applied_transpose = solve(
        gram,
        applied_basis.T,
        assume_a="sym",
        check_finite=False,
    )
    projector = np.eye(len(b)) - basis @ gram_inverse_applied_transpose
    projected_radius = np.abs(projector) @ initial_radius
    return {
        "raw_lower": projected_center - projected_radius,
        "raw_upper": projected_center + projected_radius,
        "projected_center": projected_center,
        "projected_radius": projected_radius,
        "retained_rank": selection["retained_rank"],
        "gram_condition": selection["gram_condition"],
    }


def run_intersected_projection_enclosure(
    A,
    b,
    lower,
    upper,
    iterations,
    snapshot_period,
):
    """Intersect B_0 with every B_j; retain snapshots only at the period."""
    center = 0.5 * (lower + upper)
    cg_run = scipy_cg_with_directions(A, b, center, iterations)
    directions = cg_run["directions"]
    applied_directions = cg_run["applied_directions"]

    snapshot_iterations = [0]
    running_lower_history = [lower.copy()]
    running_upper_history = [upper.copy()]
    raw_lower_history = [lower.copy()]
    raw_upper_history = [upper.copy()]
    projected_center_history = [center.copy()]
    projected_radius_history = [0.5 * (upper - lower)]
    retained_rank_history = [0]
    gram_condition_history = [1.0]

    running_lower = lower.copy()
    running_upper = upper.copy()
    for j in range(1, iterations + 1):
        raw = project_prefix_enclosure(
            A,
            b,
            lower,
            upper,
            directions[:, :j],
            applied_directions[:, :j],
        )

        # C_j = C_{j-1} intersect B_j, with C_0 = B_0.
        running_lower = np.maximum(running_lower, raw["raw_lower"])
        running_upper = np.minimum(running_upper, raw["raw_upper"])
        if np.any(running_lower > running_upper):
            raise RuntimeError(f"Running intersection became empty at j={j}.")

        if j % snapshot_period == 0 or j == iterations:
            snapshot_iterations.append(j)
            running_lower_history.append(running_lower.copy())
            running_upper_history.append(running_upper.copy())
            raw_lower_history.append(raw["raw_lower"])
            raw_upper_history.append(raw["raw_upper"])
            projected_center_history.append(raw["projected_center"])
            projected_radius_history.append(raw["projected_radius"])
            retained_rank_history.append(raw["retained_rank"])
            gram_condition_history.append(raw["gram_condition"])

    return {
        "ell": running_lower,
        "u": running_upper,
        "iterations": np.asarray(snapshot_iterations),
        "ell_history": np.asarray(running_lower_history),
        "u_history": np.asarray(running_upper_history),
        "raw_ell_history": np.asarray(raw_lower_history),
        "raw_u_history": np.asarray(raw_upper_history),
        "projected_center_history": np.asarray(projected_center_history),
        "projected_radius_history": np.asarray(projected_radius_history),
        "retained_rank_history": np.asarray(retained_rank_history),
        "gram_condition_history": np.asarray(gram_condition_history),
        "residual_norms": cg_run["residual_norms"],
        "x_cg": cg_run["x_cg"],
    }
```

```python
result = run_intersected_projection_enclosure(
    A=A_hat,
    b=b_hat,
    lower=ell_hat,
    upper=u_hat,
    iterations=cg_projection_iterations,
    snapshot_period=snapshot_period,
)

iteration_history = result["iterations"]
ell_history = result["ell_history"] / scale[None, :]
u_history = result["u_history"] / scale[None, :]
raw_ell_history = result["raw_ell_history"] / scale[None, :]
raw_u_history = result["raw_u_history"] / scale[None, :]
ell_final = result["ell"] / scale
u_final = result["u"] / scale

initial_width = u0 - ell0
width_history = u_history - ell_history
raw_intersected_width_history = (
    np.minimum(u0[None, :], raw_u_history)
    - np.maximum(ell0[None, :], raw_ell_history)
)
with np.errstate(divide="ignore", invalid="ignore"):
    shrinkage_history = np.clip(
        100.0 * (1.0 - width_history / initial_width[None, :]), 0.0, 100.0
    )
    raw_only_shrinkage_history = np.clip(
        100.0 * (1.0 - raw_intersected_width_history / initial_width[None, :]),
        0.0,
        100.0,
    )

contains_solution = np.all((ell_final <= x_star) & (x_star <= u_final))
monotone_lower = np.all(np.diff(ell_history, axis=0) >= -1e-12)
monotone_upper = np.all(np.diff(u_history, axis=0) <= 1e-12)
no_wider_than_final_v3_box = np.all(
    (u_final - ell_final) <= raw_intersected_width_history[-1] + 1e-12
)
print(f"CG iterations: {cg_projection_iterations}")
print(f"all intermediate boxes intersected: {cg_projection_iterations}")
print(f"stored animation snapshots: {len(iteration_history) - 1}")
print(f"final mean cumulative shrinkage: {np.nanmean(shrinkage_history[-1]):.2f}%")
print(f"final mean B0-intersect-Bm shrinkage: {np.nanmean(raw_only_shrinkage_history[-1]):.2f}%")
print(f"final enclosure contains x_star: {contains_solution}")
print(f"running bounds are monotone: {monotone_lower and monotone_upper}")
print(f"cumulative result is no wider than final v3-style box: {no_wider_than_final_v3_box}")
```

## Controllable cumulative-intersection animations

Each frame shows the running intersection after another 10 CG steps. The
initial box is frame zero. Coordinate indices use Python's zero-based
convention, so the `1 mod 3` coordinates are `1, 4, 7, ...`.


```python
all_indices = np.arange(len(initial_width))
mod1_indices = all_indices[all_indices % 3 == 1]


def display_shrinkage_animation(indices, title):
    frame_data = shrinkage_history[:, indices]
    finite_data = frame_data[np.isfinite(frame_data)]
    y_max = max(1.0, float(np.max(finite_data)) * 1.05) if finite_data.size else 1.0

    fig, ax = plt.subplots(figsize=(9, 4.5))
    line, = ax.plot(indices, np.nan_to_num(frame_data[0]), linewidth=1)
    label = ax.text(0.02, 0.95, "", transform=ax.transAxes, va="top")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.set(xlim=(indices[0], indices[-1]), ylim=(0, y_max))
    ax.set_xlabel("Coordinate index")
    ax.set_ylabel("Cumulative width reduction (% of original width)")
    ax.set_title(title)
    ax.grid(alpha=0.25)

    def update(frame):
        line.set_ydata(np.nan_to_num(frame_data[frame]))
        label.set_text(f"CG iteration {iteration_history[frame]}")
        return line, label

    animation = FuncAnimation(fig, update, frames=len(iteration_history), interval=180)
    display(HTML(animation.to_jshtml(fps=5, default_mode="loop")))
    plt.close(fig)
    return animation


def display_bounds_animation(indices, title):
    lower_data = ell_history[:, indices]
    upper_data = u_history[:, indices]
    y_min = float(np.min(lower_data))
    y_max = float(np.max(upper_data))
    padding = 0.03 * max(y_max - y_min, 1.0)

    fig, ax = plt.subplots(figsize=(9, 4.5))
    lower_line, = ax.plot(indices, lower_data[0], label="running lower", linewidth=1)
    upper_line, = ax.plot(indices, upper_data[0], label="running upper", linewidth=1)
    label = ax.text(0.02, 0.95, "", transform=ax.transAxes, va="top")
    ax.set(xlim=(indices[0], indices[-1]), ylim=(y_min - padding, y_max + padding))
    ax.set_xlabel("Coordinate index")
    ax.set_ylabel("Bound value")
    ax.set_title(title)
    ax.grid(alpha=0.25)
    ax.legend()

    def update(frame):
        lower_line.set_ydata(lower_data[frame])
        upper_line.set_ydata(upper_data[frame])
        label.set_text(f"CG iteration {iteration_history[frame]}")
        return lower_line, upper_line, label

    animation = FuncAnimation(fig, update, frames=len(iteration_history), interval=180)
    display(HTML(animation.to_jshtml(fps=5, default_mode="loop")))
    plt.close(fig)
    return animation
```

```python
all_coordinate_animation = display_shrinkage_animation(
    all_indices,
    "Running-intersection shrinkage along all coordinates",
)
```

```python
mod1_shrinkage_animation = display_shrinkage_animation(
    mod1_indices,
    "Running-intersection shrinkage along coordinates 1 mod 3",
)
```

```python
mod1_bounds_animation = display_bounds_animation(
    mod1_indices,
    "Running lower and upper bounds for coordinates 1 mod 3",
)
```
