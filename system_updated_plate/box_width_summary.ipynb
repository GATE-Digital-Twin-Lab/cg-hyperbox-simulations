```python
from pathlib import Path

import numpy as np
import pandas as pd

data_dir = Path.cwd()
if not list(data_dir.glob("box_*.dat")):
    data_dir = Path("system_updated_plate")

rows = []
for path in sorted(data_dir.glob("box_*.dat"), key=lambda item: int(item.stem.split("_")[-1])):
    box = np.loadtxt(path, dtype=float)
    widths = box[:, 1] - box[:, 0]
    zero_count = int(np.count_nonzero(widths == 0.0))
    total_count = int(widths.size)
    nonzero_widths = widths[widths != 0.0]
    rows.append({
        "box": path.stem,
        "zero-width coordinates": f"{zero_count:,} / {total_count:,} ({100.0 * zero_count / total_count:.2f}%)",
        "median non-zero width": float(np.median(nonzero_widths)) if nonzero_widths.size else np.nan,
        "biggest width": float(np.max(widths)),
    })

summary = pd.DataFrame(rows).set_index("box")
summary.style.format({
    "median non-zero width": "{:.6e}",
    "biggest width": "{:.6e}",
})
```
